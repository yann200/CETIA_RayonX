Ce notebook permet de:
- retirer la distorsion des imges
- modifier l'extension des images prises au rayon X en png

```python
import os
import cv2

def crop_y_distortion_batch(input_folder, output_folder, scale_y):
    os.makedirs(output_folder, exist_ok=True)

    for filename in os.listdir(input_folder):
            input_path = os.path.join(input_folder, filename)
            output_path = os.path.join(output_folder, filename)

            # Lire l'image
            img = cv2.imread(input_path, cv2.IMREAD_UNCHANGED)
            if img is None:
                print(f"[Erreur] Impossible de lire : {filename}")
                continue

            h, w = img.shape[:2]
            new_h = int(h * scale_y)

            # Crop sur la hauteur (on garde le haut)
            cropped = img[:new_h, :]

            # Sauvegarde
            cv2.imwrite(output_path, cropped)
            print(f"[OK] {filename} -> {output_path}")

# Paramètres
scale_y = 280 / 308  # à adapter à tes dimensions
input_folder = "Pant_test"
output_folder = "Pant_test_undistorted"

crop_y_distortion_batch(input_folder, output_folder, scale_y)
```

```python
import os
import cv2

def batch_tiff_to_png(input_folder, output_folder):
    os.makedirs(output_folder, exist_ok=True)

    for filename in os.listdir(input_folder):
        if filename.lower().endswith((".tif", ".tiff")):
            input_path = os.path.join(input_folder, filename)
            base_name = os.path.splitext(filename)[0]  # sans extension
            output_path = os.path.join(output_folder, base_name + ".png")

            img = cv2.imread(input_path, cv2.IMREAD_UNCHANGED)
            if img is None:
                print(f"[Erreur] Impossible de lire : {filename}")
                continue

            cv2.imwrite(output_path, img)
            print(f"[OK] {filename} -> {output_path}")

# Exemple d'utilisation :
input_folder = "Pant_test_undistorted"
output_folder = "Pant_test_undistorted_png"

batch_tiff_to_png(input_folder, output_folder)
```

```python
# convertions des images 16 bits → 8 bits avec un ajustement du contraste (normalisation).
import numpy as np
import cv2

def convert_png_to_8bit_folder(input_dir, output_dir):
    """
    Convertit toutes les images PNG (potentiellement 16 bits ou float) d'un dossier en PNG 8 bits.
    """
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    for filename in os.listdir(input_dir):
        if filename.lower().endswith(".png"):
            input_path = os.path.join(input_dir, filename)
            output_path = os.path.join(output_dir, filename)

            img = cv2.imread(input_path, cv2.IMREAD_UNCHANGED)
            if img is None:
                print(f"❌ Impossible de lire : {input_path}")
                continue

            # Vérifie si l'image est déjà en 8 bits
            if img.dtype == np.uint8:
                cv2.imwrite(output_path, img)
                print(f"✅ Déjà 8 bits : Copié {filename}")
                continue

            # Sinon, normalise pour convertir vers 8 bits
            img_8bit = cv2.normalize(img, None, 0, 255, cv2.NORM_MINMAX)
            img_8bit = img_8bit.astype(np.uint8)
            cv2.imwrite(output_path, img_8bit)
            print(f"✅ Converti en 8 bits : {filename}")

# 🔽 Appel de la fonction (en dehors de la fonction elle-même)
input_dir = "Pant_test_undistorted_png"
output_dir = "Pant_test_undistorted_png_8bit"
convert_png_to_8bit_folder(input_dir, output_dir)
```
